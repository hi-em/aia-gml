# OBJ to BREP — Police Station Ground Floor

## //01 Export OBJ to BREP
> Convert the ground-floor mesh to a raw topologic BREP.

```python
from topologicpy.Topology import Topology
from topologicpy.Cluster  import Cluster

OBJ_PATH  = r"../assets/gf-floor-plan.obj"
BREP_PATH = r"../assets/gf-floor-plan.brep"

topo = Topology.ByOBJPath(OBJ_PATH)
if isinstance(topo, list):
    topo = Cluster.ByTopologies(topo)

print("Type:", Topology.TypeAsString(topo), "Faces:", len(Topology.Faces(topo)))

ok = Topology.ExportToBREP(topo, path=BREP_PATH, overwrite=True)
print("Exported:", ok)
```

## //02 Simplify BREP | Extract Floor Face
> Load BREP | merge triangles into shell | derive outer boundary | export clean single face.

```python
from topologicpy.Topology import Topology
from topologicpy.Cluster  import Cluster
from topologicpy.Shell    import Shell
from topologicpy.Face     import Face

BREP_IN  = r"../assets/gf-floor-plan.brep"
BREP_OUT = r"../assets/gf-floor-plan-face.brep"

floor_plan = Topology.ByBREPPath(BREP_IN)
print("Loaded:", floor_plan)

triangles = Cluster.Faces(floor_plan)
shell     = Shell.ByFaces(triangles)
eb        = Shell.ExternalBoundary(shell)
ib_list   = Shell.InternalBoundaries(shell)
new_face  = Face.ByWires(eb, ib_list)
new_face  = Topology.RemoveCollinearEdges(new_face)

print(f"Final edge count: {len(Topology.Edges(new_face))}")
print(new_face)

Topology.Show(new_face)
Topology.ExportToBREP(new_face, path=BREP_OUT, overwrite=True)
```
